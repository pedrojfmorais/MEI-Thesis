```python
import json
import subprocess
from IPython.display import Image, display


def analyze(dataset_name, logs_file, algorithms):
  cmd = [
    "python",
    "/content/drive/MyDrive/mad/code/SPL-MAD/EER Analysis/analyze_logs.py",
    dataset_name,
    logs_file,
    *algorithms
  ]

  result = subprocess.run(
      cmd,
      capture_output=True,
      text=True
  )

  print(result.stdout)
  print(result.stderr)

  display(Image(filename=f"eer_trends_{dataset_name}.png"))
```

```python
analyze(
    dataset_name="SMDD",
    logs_file="/content/drive/MyDrive/mad/code/SPL-MAD/EER Analysis/logs_testing_smdd.txt",
    algorithms=["Non-Cropped", "Cropped"]
)
```

```python
analyze(
    dataset_name="FRLL-Morphs",
    logs_file="/content/drive/MyDrive/mad/code/SPL-MAD/EER Analysis/logs_testing_frll-morphs.txt",
    algorithms=["AMSL", "FaceMorpher", "OpenCV", "StyleGAN2", "WebMorph"]
)
```
